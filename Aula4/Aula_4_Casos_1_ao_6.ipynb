```python
import os
from dotenv import load_dotenv
from openai import OpenAI
import numpy as np
import pandas as pd
import tiktoken
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_text_splitters import CharacterTextSplitter
from pathlib import Path
from sklearn.metrics.pairwise import cosine_similarity, cosine_distances
import math
```

```python
def gerar_embeddings(termos):

  load_dotenv()
  client = OpenAI(
      api_key=''
      )

  embeddings = []

  response = client.embeddings.create(
      model="text-embedding-3-small", input=[str(t) for t in termos]
  )

  embeddings = [item.embedding for item in response.data]

  return np.array(embeddings)
```

```python
def teste_recursivo_metadado(caminho, chunk_size, chunk_overlap):
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=chunk_size,
        chunk_overlap=chunk_overlap,
        add_start_index=True
    )

    with open(caminho, 'r', encoding='utf-8') as f:
        conteudo_do_texto = f.read()

    resultado = text_splitter.create_documents(
        texts=[conteudo_do_texto],
        metadatas=[{"fonte": caminho}]
    )

    return resultado
```

```python
def buscar_arquivos():

    pasta = Path(r"/content/")
    caminhos = list(pasta.glob("*.md"))
    return caminhos
```

```python
resultados_dados_metadados = {}

def gerar_base_dados_metadados(parametros):
  for i in range(len(parametros)):
      chave = (str(parametros[i][0]) + " -- " + str(parametros[i][1]) + " -- " + str(parametros[i][2]))

      resultados_dados_metadados[chave] = teste_recursivo_metadado(
          parametros[i][0],
          parametros[i][1],
          parametros[i][2]
      )
```

```python
def info_recursivo(caminho, chunk_size, chunk_overlap):
  bloco = []
  bloco = teste_recursivo_metadado(caminho, chunk_size, chunk_overlap)
  print("chunk_size", chunk_size, "chunk_overlap", chunk_overlap)

  # Total de chunks
  print("Total de chunks: " + str(len(bloco)))

  # Tamanho medio chunck
  tamanho_medio = sum(len(s.page_content) for s in bloco) / len(bloco)
  print("tamanho médio: " + str(tamanho_medio))

  # Menor chunks (Correção com lambda)
  menor_documento = min(bloco, key=lambda s: len(s.page_content))
  print("menor string: " + str(len(menor_documento.page_content)))

  # Maior chunks (Correção com lambda)
  maior_documento = max(bloco, key=lambda s: len(s.page_content))
  print("maior string: " + str(len(maior_documento.page_content)))

  # Numeros de tokens
  print("total de tokens: " + str(contar_tokens_dos_blocos(bloco)))
```

```python
def contar_tokens_dos_blocos(lista_de_blocos, modelo="text-embedding-3-small"):
    
    codificador = tiktoken.encoding_for_model(modelo)

    total_tokens = 0
    tokens_por_bloco = []

    for bloco in lista_de_blocos:
        
        texto_puro = bloco.page_content

        num_tokens = len(codificador.encode(texto_puro, disallowed_special=()))
        tokens_por_bloco.append(num_tokens)
        total_tokens += num_tokens

    media_tokens = total_tokens / len(lista_de_blocos) if lista_de_blocos else 0

    return total_tokens
```

```python
import os
import json

def criar_json(chuncks, embeddings, metadados, nome_parametros):
  dados_json = [
      {
          "Texto": texto,
          "Embeddings": emb.tolist() if hasattr(emb, "tolist") else emb,
          "Metadados": meta.metadata if hasattr(meta, "metadata") else meta
      }
      for texto, emb, meta in zip(chuncks, embeddings, metadados)
  ]

  pasta_destino = "resultados"

  os.makedirs(pasta_destino, existindo_ok=True)

  nome = os.path.path.join(pasta_destino, f"{nome_parametros}.json")

  with open(nome, 'w', encoding='utf-8') as f:
      json.dump(dados_json, f, indent=4, ensure_ascii=False, default=str)

  print(f"JSON gerado com sucesso em: {nome}")
```
